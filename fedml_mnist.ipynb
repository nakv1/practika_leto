```python
import sys
import fedml
from fedml import FedMLRunner

def run_fedml(config_path):
    sys.argv = ["fedml_mnist.py", "--cf", config_path]
    args = fedml.init()
    device = fedml.device.get_device(args)
    dataset, output_dim = fedml.data.load(args)
    model = fedml.model.create(args, output_dim)
    FedMLRunner(args, device, dataset, model).run()

print("Функция запуска FedML готова")
```

```python
run_fedml("fedml_config_lr5.yaml")
```

```python
run_fedml("fedml_config_cnn.yaml")
```
